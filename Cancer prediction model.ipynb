```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('data.csv')
```

```python
df.head()
```

```python
df=df.drop('Unnamed: 32', axis=1)
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['diagnosis'].unique()
```

```python
dummies=pd.get_dummies(df['diagnosis'], drop_first=True,dtype=int)
```

```python
df=pd.concat([df,dummies], axis=1)
```

```python
df.head()
```

```python
df=df.drop(['diagnosis', 'id'], axis=1)
```

```python
sns.set_style('darkgrid')
```

```python
sns.histplot(df['concave points_mean'], kde=True )
```

```python
a=df.corr()
```

```python
plt.figure(figsize=(20,15))
sns.heatmap(a, cmap='rocket', annot=True, linewidth=2)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X=df.drop('M', axis=1)
y=df['M']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=101)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
x_train_scaled=scaler.fit_transform(X_train)
x_test_scaled=scaler.transform(X_test)
```

### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression(solver='lbfgs',max_iter=5000)
```

```python
lr.fit(x_train_scaled,y_train)
```

```python
predictions=lr.predict(x_test_scaled)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

```python
print(confusion_matrix(y_test, predictions))
```

```python
print(classification_report(y_test, predictions))
```

```python
print(accuracy_score(y_test, predictions))
```

### K-Nearest Neighbour

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier()
```

```python
knn.fit(x_train_scaled,y_train)
```

```python
kpredictions=knn.predict(x_test_scaled)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

```python
print(confusion_matrix(y_test, kpredictions))
```

```python
print(classification_report(y_test, kpredictions))
```

```python
print(accuracy_score(y_test, kpredictions))
```

### Support Vector Machines

```python
from sklearn.svm import SVC
```

```python
sc=SVC()
```

```python
sc.fit(x_train_scaled, y_train)
```

```python
scpredictions=sc.predict(x_test_scaled)
```

```python
print(confusion_matrix(y_test, scpredictions))
```

```python
print(classification_report(y_test, scpredictions))
```

### Grid_Search Optimized Support Vector Machine

```python
from sklearn.model_selection import GridSearchCV
```

```python
paradm_grid={'C':[0.1,1,10,100,1000],
            'gamma':[1,0.1,0.01,0.001,0.0001],
             'kernel':['rbf']}
```

```python
grid=GridSearchCV(SVC(), paradm_grid, refit=True, verbose=5)
```

```python
grid.fit(x_train_scaled,y_train)
```

```python
grid.best_params_
```

```python
grid.best_estimator_
```

```python
grid_predictions=grid.predict(x_test_scaled)
```

```python
print(confusion_matrix(y_test,grid_predictions))
```

```python
print(classification_report(y_test, grid_predictions))
```

### RandomForest


```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc=RandomForestClassifier(n_estimators=100)
```

```python
rfc.fit(x_train_scaled,y_train)
```

```python
rfcpredictions=rfc.predict(x_test_scaled)
```

```python
print(confusion_matrix(y_test, rfcpredictions))
```

```python
print(classification_report(y_test, rfcpredictions))
```

### The default SVC model achieved a test accuracy of 98% , outperfoming the GridSearchC-optimized model, which achieved 96%. Therefore the default SVC was selected as the final model based on its perfomance on unseen test set
